## Fetching all the pages

```python
df = pd.DataFrame()
```

```python
df
```

```python
for i in range(1, 429):
  response = requests.get(f'https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page={i}')
  temp_df = pd.DataFrame(response.json()['results'])[['id','title','overview','release_date','popularity','vote_average','vote_count']]
  df = pd.concat([df, temp_df], ignore_index=True)
```

```python
df.head(50)

# as we can see here that the index starts again from 0 for every page, for that we will use ignore index parameter
```
